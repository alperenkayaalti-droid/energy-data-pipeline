# Energy Consumption Analysis & Forecasting
This notebook performs Exploratory Data Analysis (EDA) and trend smoothing on the cleaned energy data processed by our ETL pipeline.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
# Load the processed daily KPIs
# If the file doesn't exist yet, we generate the mock data visualization structure
data_path = '../data/daily_energy_kpis.csv'
if os.path.exists(data_path):
    df = pd.read_csv(data_path)
else:
    dates = pd.date_range(start="2026-01-01", end="2026-05-31", freq="D")
    np.random.seed(42)
    base_electricity = np.random.normal(loc=28800, scale=1200, size=len(dates))
    df = pd.DataFrame({
        'date': dates,
        'electricity_kwh': base_electricity
    })
    df['electricity_7d_moving_avg'] = df['electricity_kwh'].rolling(window=7, min_periods=1).mean()

df['date'] = pd.to_datetime(df['date'])
```

```python
# Visualize the Industrial Electricity Consumption Trend
plt.figure(figsize=(12, 5))
plt.plot(df['date'], df['electricity_kwh'], label='Daily Electricity Consumption (kWh)', color='#2ecc71', alpha=0.6)
plt.plot(df['date'], df['electricity_7d_moving_avg'], label='7-Day Moving Average (Trend)', color='#e67e22', linewidth=2, linestyle='--')

plt.title('Factory Daily Electricity Consumption & Baseline Trend', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Timeline', fontsize=12)
plt.ylabel('Electricity (kWh)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
